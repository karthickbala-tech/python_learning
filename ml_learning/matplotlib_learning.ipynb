```python
#object oriented approach
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Sample")
ax.set_xlabel("Data")
ax.set_ylabel("Values")
plt.show()
```

```python
cars = {
    'bmw': 1050000,
    'audi': 2137622,
    'benz': 4654464,
    'toyota': 850000,
    'honda': 720000,
    'ford': 950000,
    'nissan': 680000,
    'chevrolet': 1020000,
    'volkswagen': 1150000,
    'kia': 630000,
    'hyundai': 700000,
    'mazda': 780000,
    'subaru': 820000,
    'jaguar': 3200000,
    'land_rover': 4500000,
    'tesla': 3800000,
    'porsche': 5500000,
    'ferrari': 12000000,
    'lamborghini': 15000000,
    'mercedes_gla': 3200000,
    'volvo': 900000,
    'mitsubishi': 650000,
    'suzuki': 580000,
    'lexus': 2600000,
    'acura': 2400000,
    'infiniti': 2100000,
    'mini': 800000,
    'peugeot': 720000,
    'renault': 690000,
    'citroen': 640000
}
```

```python
fig , ax=plt.subplots()
ax.plot(cars.keys(),cars.values())
plt.xticks(rotation=90)
plt.show()
```

```python
#line plot
fig, ax = plt.subplots(figsize=(12,6))
ax.plot(cars.keys(), cars.values(), marker='o', color='r')
ax.set_title("Car Prices - Line Plot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
#scatter plot
fig, ax = plt.subplots(figsize=(12,6))
ax.scatter(cars.keys(), cars.values(), color='blue')
ax.set_title("Car Prices - Scatter Plot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#HISTOGRAM
prices = list(cars.values())

plt.figure(figsize=(10,6))
plt.hist(prices, bins=8, color='skyblue', edgecolor='black')
plt.title("Distribution of Car Prices")
plt.xlabel("Price Range")
plt.ylabel("Number of Cars")
plt.show()
```

```python
#bar plot 
fig, ax = plt.subplots(figsize=(12,6))
ax.bar(cars.keys(), cars.values(), color='skyblue')
ax.set_title("Car Prices - Bar Plot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
#Horizontal Bar Plot
fig, ax = plt.subplots(figsize=(12,6))
ax.barh(list(cars.keys()), list(cars.values()), color='orange')
ax.set_title("Car Prices - Horizontal Bar Plot")
plt.tight_layout()
plt.show()
```

```python
#Step Plot
fig, ax = plt.subplots(figsize=(12,6))
ax.step(cars.keys(), cars.values(), where='mid', color='green')
ax.set_title("Car Prices - Step Plot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
#Horizontal Bar PlotHorizontal Bar PlotHorizontal Bar Plot
ax.fill_between(cars.keys(), cars.values(), color='purple', alpha=0.3)
ax.plot(cars.keys(), cars.values(), color='green')
plt.show()
```

```python
#AREA \FILL BETWEEN PLOT
fig, ax = plt.subplots(figsize=(12,6))
ax.fill_between(cars.keys(), cars.values(), color='purple', alpha=0.3)
ax.plot(cars.keys(), cars.values(), color='green')
ax.set_title("Car Prices - Area Plot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12,6))
ax.set_title("Car Prices")
ax.set_xlabel("Car Brand")
ax.set_ylabel("Price (in currency units)")
plt.xticks(rotation=45, ha='right')  # rotate x labels for readability
plt.tight_layout()  # adjust layout so labels fit
plt.show()
```

```python
# Create a histogram
prices = list(cars.values())
plt.figure(figsize=(10,6))
plt.hist(prices, bins=10, color='skyblue', edgecolor='black')  # bins = how many groups
plt.title("Distribution of Car Prices")
plt.xlabel("Price Range")
plt.ylabel("Number of Cars")
plt.show()
```

```python
fig, ax = plt.subplots(3, figsize=(12, 15)) 

ax[0].plot(cars.keys(), cars.values(), marker='o', color='r')
ax[0].set_title("Line Plot")
ax[0].set_ylabel("Price")
ax[0].tick_params(axis='x', rotation=45)

ax[1].bar(cars.keys(), cars.values(), color='g')
ax[1].set_title("Bar Plot")
ax[1].set_ylabel("Price")
ax[1].tick_params(axis='x', rotation=45)

ax[2].scatter(cars.keys(), cars.values(), color='b', marker='o')
ax[2].set_title("Scatter Plot")
ax[2].set_xlabel("Car Brand")
ax[2].set_ylabel("Price")
ax[2].tick_params(axis='x', rotation=45)

plt.tight_layout()  
plt.show()
```

```python
import pandas as pd

csv_file = pd.read_csv("/home/parrot/Documents/machine_learning/datasets/dunya.csv")

x = csv_file['Country']
y1 = csv_file['Revenue']
y2 = csv_file['Nüfus']
y3 = csv_file['Per person']

plt.style.use('seaborn-v0_8')

fig, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, figsize=(20, 12))
ax1.bar(x, y1, color='tab:blue', label='Revenue')
ax1.set_ylabel('Revenue')
ax1.legend()

ax2.bar(x, y2, color='tab:orange', label='Population')
ax2.set_ylabel('Population')
ax2.legend()

ax3.bar(x, y3, color='tab:green', label='Per Person')
ax3.set_ylabel('Per Person')
ax3.set_xlabel('Country')
ax3.legend()

plt.xticks(rotation=45, ha='right')
plt.suptitle("Income Comparison by Country", fontsize=18)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, figsize=(20, 12))

ax1.bar(x, y1, color='tab:blue', label='Revenue')
ax2.bar(x, y2, color='tab:orange', label='Population')
ax3.bar(x, y3, color='tab:green', label='Per Person')

fig.legend(
    loc='upper right',
    bbox_to_anchor=(0.95, 0.95),
    title="Metrics"
)
ax1.set_ylabel('Revenue')
ax2.set_ylabel('Population')
ax3.set_ylabel('Per Person')
ax3.set_xlabel('Country')

plt.xticks(rotation=45, ha='right')
plt.suptitle("Income Comparison by Country", fontsize=18)
plt.tight_layout(rect=[0, 0, 0.9, 0.96])
plt.show()
```
