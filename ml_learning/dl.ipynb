```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

# ------------------ Create Figure ------------------
fig, axs = plt.subplots(1, 3, figsize=(24, 8))
plt.subplots_adjust(left=0.05, right=0.95, top=0.9, bottom=0.1, wspace=0.4)

# Make all axes fill their space
for ax in axs:
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 6)
    ax.axis('off')

# ------------------ ANN ------------------
axs[0].set_title('ANN (Artificial Neural Network)', fontsize=14)
neurons = [3, 4, 2]  # number of neurons in each layer
for i, n_neurons in enumerate(neurons):
    for n in range(n_neurons):
        axs[0].add_patch(Circle((i*2, n+1), 0.2, color='skyblue', ec='black'))

# Connect neurons
for i, n1 in enumerate(neurons[:-1]):
    for n1_idx in range(n1):
        for n2_idx in range(neurons[i+1]):
            axs[0].plot([i*2, (i+1)*2], [n1_idx+1, n2_idx+1], 'k-', lw=0.5)

# ------------------ CNN ------------------
axs[1].set_title('CNN (Convolutional Neural Network)', fontsize=14)
layers = [
    ('Input Image', 'lightcoral'),
    ('Convolution', 'lightgreen'),
    ('Pooling', 'lightblue'),
    ('Fully Connected', 'orange'),
    ('Output', 'purple')
]

x = 0
for name, color in layers:
    axs[1].add_patch(Rectangle((x, 1.5), 1.5, 3, facecolor=color, edgecolor='black'))
    axs[1].text(x+0.75, 5, name, ha='center', fontsize=10)
    x += 2  # spacing between layers

# ------------------ RNN ------------------
axs[2].set_title('RNN (Recurrent Neural Network)', fontsize=14)
seq_len = 4
# Input sequence
for t in range(seq_len):
    axs[2].add_patch(Rectangle((t*1.8,1), 1,1, facecolor='lightcoral', edgecolor='black'))
    axs[2].text(t*1.8+0.5, 2.2, f'X{t+1}', ha='center', fontsize=10)

# Recurrent arrows
for t in range(seq_len-1):
    axs[2].arrow(t*1.8+0.9, 1.5, 0.8, 0, head_width=0.1, head_length=0.1, fc='black', ec='black')

# Output sequence
for t in range(seq_len):
    axs[2].add_patch(Rectangle((t*1.8,3),1,1, facecolor='purple', edgecolor='black'))
    axs[2].text(t*1.8+0.5, 4, f'Y{t+1}', ha='center', fontsize=10)

# ------------------ Show Figure ------------------
plt.show()
```
